# Bouncing ball — visual inspection

**Vérifie que la balle occupe bien ~15–20 % de l'image AVANT de lancer la collecte et l'entraînement.**

Actions (`action_dim=1`) : `+1` = accélérer le long de la vitesse, `-1` = ralentir. Tirage aléatoire ±1 à chaque step.

Guide des rayons sur 32×32 :
- `ball_radius=5` → ~8 %
- `ball_radius=7` → ~15 %
- `ball_radius=8` → ~19 %

```python
import sys
sys.path.insert(0, "..")

from bouncing_ball import BouncingBallConfig, preview_episode

cfg = BouncingBallConfig(
    img_size=32,
    ball_radius=7,  # ~15% of image
    initial_speed=2.5,
)

preview_episode(episode_len=20, seed=42, cfg=cfg)
```

```python
# GIF — elastic bounces + random ±1 throttle actions
from bouncing_ball import BouncingBallConfig, save_episode_gif

cfg = BouncingBallConfig(img_size=32, ball_radius=7, action_dim=1)
save_episode_gif("../outputs/bouncing_ball_elastic.gif", episode_len=40, seed=42, cfg=cfg)
```

```python
# Pixel stats on a single frame (nearest-neighbor = true pixels)
import numpy as np
import matplotlib.pyplot as plt
from bouncing_ball import roll_out_episode, ball_pixel_stats

rng = np.random.default_rng(42)
obs, actions = roll_out_episode(20, rng, cfg)
t = 10
n_px, frac = ball_pixel_stats(obs[t], cfg)
print(f"Frame t={t}: {n_px} ball pixels ({frac:.2f}% of {cfg.img_size}x{cfg.img_size})")

fig, ax = plt.subplots(1, 1, figsize=(4, 4))
ax.imshow(obs[t].transpose(1, 2, 0), interpolation="nearest")
ax.set_title(f"t={t} | {n_px}px ({frac:.1f}%)")
ax.axis("off")
plt.show()
```
